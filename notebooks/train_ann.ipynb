# ANN Training – Diabetes Risk Prediction

**Dataset:** Pima Indians Diabetes Dataset (Kaggle)  
**Model:** Artificial Neural Network (TensorFlow/Keras)  
**Task:** Binary Classification – Diabetic / Not Diabetic

---

## Step 1 – Download Dataset via Kaggle CLI

Place your `kaggle.json` in `C:\Users\<YourName>\.kaggle\kaggle.json`, then run:

```bash
kaggle datasets download -d uciml/pima-indians-diabetes-database -p datasets --unzip
```

````python
import os, sys
sys.path.insert(0, os.path.abspath('..'))

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import tensorflow as tf

print('TensorFlow:', tf.__version__)

df = pd.read_csv('../datasets/diabetes.csv')
print('Shape:', df.shape)
df.head()
````

````python
# ── EDA ────────────────────────────────────────────────────────
print('Missing values:\n', df.isnull().sum())
print('\nClass distribution:\n', df['Outcome'].value_counts())
df.describe()
````

````python
# ── Train using app utility ────────────────────────────────────
from utils.ann_utils import train_ann

metrics = train_ann(df, target_col='Outcome', epochs=50)

print(f"Accuracy  : {metrics['accuracy']*100:.2f}%")
print(f"Precision : {metrics['precision']*100:.2f}%")
print(f"Recall    : {metrics['recall']*100:.2f}%")
print(f"F1 Score  : {metrics['f1']*100:.2f}%")
print(f"ROC AUC   : {metrics['roc_auc']:.4f}")
````

````python
# ── Training Curves ────────────────────────────────────────────
fig, axes = plt.subplots(1, 2, figsize=(14, 4))

axes[0].plot(metrics['train_acc'], label='Train')
axes[0].plot(metrics['val_acc'], label='Validation', linestyle='--')
axes[0].set_title('Accuracy'); axes[0].legend(); axes[0].grid(True, alpha=0.3)

axes[1].plot(metrics['train_loss'], label='Train', color='red')
axes[1].plot(metrics['val_loss'], label='Validation', linestyle='--', color='orange')
axes[1].set_title('Loss'); axes[1].legend(); axes[1].grid(True, alpha=0.3)

plt.tight_layout()
plt.show()
````

````python
# ── Confusion Matrix ───────────────────────────────────────────
import plotly.express as px

fig = px.imshow(metrics['cm'], text_auto=True, color_continuous_scale='Blues',
                x=['Negative','Positive'], y=['Negative','Positive'],
                labels=dict(x='Predicted', y='Actual'),
                title='Confusion Matrix')
fig.update_layout(template='plotly_white')
fig.show()
````
